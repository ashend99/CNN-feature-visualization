```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
import numpy as np
from models import SimpleCNN, trainset, testset, trainloader, testloader
```

```python
model = SimpleCNN()
model.load_state_dict(torch.load("../models/cnn_cifar10.pth"))
model.eval()
```

```python
activations = {}

def hook_fn(module, input, output):
    activations['conv1'] = output

# Add a hook to trigger when the first convolutional layer is executed (train or inference)
model.features[0].register_forward_hook(hook_fn)
```

```python
image, label = testset[0]   # Get the first image and its label
input_img = image.unsqueeze(0)  # Add batch dimension

output = model(input_img)   # Forward pass to trigger the hook and store activations
```

```python
# Visualize the feature maps from the first convolutional layer
fm = activations['conv1'][0]

fig, axes = plt.subplots(4, 4, figsize=(6,6))

for i, ax in enumerate(axes.flat):
    ax.imshow(fm[i].detach(), cmap='viridis')
    ax.axis('off')

plt.show()
```
